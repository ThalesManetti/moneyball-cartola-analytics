```python
# Setup
import sys
sys.path.append('../src')

from statsbombpy import sb
import pandas as pd
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

from utils.config import STATSBOMB_COMPS, BRONZE_STATSBOMB
from utils.logger import setup_logger

logger = setup_logger('statsbomb', 'logs/statsbomb.log')

print("✅ Setup completo!")
print(f"📦 statsbombpy version: {sb.__version__ if hasattr(sb, '__version__') else 'N/A'}")
print(f"🌍 Competições: {STATSBOMB_COMPS}")
```

```python
print("🔍 Buscando competições disponíveis no StatsBomb...\n")

# Buscar todas as competições
competitions = sb.competitions()

print(f"✅ {len(competitions)} competições disponíveis\n")
print("📋 Primeiras 10 competições:")
print(competitions[['competition_id', 'season_id', 'competition_name', 'season_name', 'country_name']].head(10))

# Filtrar competições que queremos
comps_interest = competitions[
    competitions['competition_name'].isin(STATSBOMB_COMPS)
]

print(f"\n🎯 Competições selecionadas: {len(comps_interest)}\n")
print(comps_interest[['competition_id', 'season_id', 'competition_name', 'season_name']])

logger.info(f"Competições disponíveis: {len(comps_interest)}")
```

```python
# Extraindo partidas
print("="*70)
print(" "*20 + "⚽ EXTRAINDO PARTIDAS")
print("="*70 + "\n")

all_matches = []

for _, comp in tqdm(comps_interest.iterrows(), total=len(comps_interest), desc="Competições"):
    
    comp_id = comp['competition_id']
    season_id = comp['season_id']
    comp_name = comp['competition_name']
    season_name = comp['season_name']
    
    try:
        # Buscar partidas
        matches = sb.matches(competition_id=comp_id, season_id=season_id)
        
        if not matches.empty:
            matches['competition_name'] = comp_name
            matches['season_name'] = season_name
            all_matches.append(matches)
            
            print(f"✅ {comp_name} {season_name}: {len(matches)} partidas")
            logger.info(f"{comp_name} {season_name}: {len(matches)} partidas")
    
    except Exception as e:
        print(f"❌ Erro em {comp_name} {season_name}: {e}")
        logger.error(f"Erro em {comp_name} {season_name}: {e}")

# Consolidar
if all_matches:
    df_matches = pd.concat(all_matches, ignore_index=True)
    
    print(f"\n{'='*70}")
    print(f"✅ Total de partidas extraídas: {len(df_matches):,}")
    print(f"📊 Competições: {df_matches['competition_name'].nunique()}")
    print(f"📅 Temporadas: {df_matches['season_name'].nunique()}")
    print(f"{'='*70}")
    
    # Mostrar distribuição
    print("\n📊 Partidas por competição:")
    print(df_matches['competition_name'].value_counts())
    
else:
    df_matches = pd.DataFrame()
    print("❌ Nenhuma partida extraída")
```

```python
# Extraindo eventos de partidas (amostra)

print("="*70)
print(" "*15 + "📊 EXTRAINDO EVENTOS (AMOSTRA)")
print("="*70 + "\n")

# Extrair eventos apenas de uma amostra (10 partidas por competição)
# Eventos completos são MUITO pesados (>1GB)

sample_matches = df_matches.groupby('competition_name').head(10)

print(f"📊 Extraindo eventos de {len(sample_matches)} partidas (amostra)...\n")

all_events = []
errors = 0

for match_id in tqdm(sample_matches['match_id'].values, desc="Eventos"):
    
    try:
        events = sb.events(match_id=match_id)
        
        if not events.empty:
            events['match_id'] = match_id
            all_events.append(events)
    
    except Exception as e:
        errors += 1
        logger.warning(f"Erro em match {match_id}: {e}")

# Consolidar
if all_events:
    df_events = pd.concat(all_events, ignore_index=True)
    
    print(f"\n✅ Eventos extraídos: {len(df_events):,}")
    print(f"⚽ Partidas processadas: {len(sample_matches) - errors}")
    print(f"❌ Erros: {errors}")
    
    print(f"\n📊 Tipos de eventos:")
    print(df_events['type'].value_counts().head(10))
    
else:
    df_events = pd.DataFrame()
    print("❌ Nenhum evento extraído")
```

```python
# Extraindo lineups
print("="*70)
print(" "*20 + "👥 EXTRAINDO LINEUPS")
print("="*70 + "\n")

all_lineups = []

# Pegar amostra de partidas
sample_for_lineups = df_matches.head(50)  # Primeiras 50 partidas

print(f"📊 Extraindo lineups de {len(sample_for_lineups)} partidas...\n")

for match_id in tqdm(sample_for_lineups['match_id'].values, desc="Lineups"):
    
    try:
        lineups = sb.lineups(match_id=match_id)
        
        # LineUps retorna dict com 2 times
        for team_name, players in lineups.items():
            if not players.empty:
                players['match_id'] = match_id
                players['team_name'] = team_name
                all_lineups.append(players)
    
    except Exception as e:
        logger.warning(f"Erro em lineup match {match_id}: {e}")

# Consolidar
if all_lineups:
    df_lineups = pd.concat(all_lineups, ignore_index=True)
    
    print(f"\n✅ Jogadores em lineups: {len(df_lineups):,}")
    print(f"⚽ Partidas: {df_lineups['match_id'].nunique()}")
    print(f"👥 Jogadores únicos: {df_lineups['player_name'].nunique()}")
    
else:
    df_lineups = pd.DataFrame()
    print("❌ Nenhum lineup extraído")
```

```python
# Salvando dados - Bronze
print("\n" + "="*70)
print(" "*20 + "💾 SALVANDO DADOS - BRONZE")
print("="*70 + "\n")

# Criar diretório
BRONZE_STATSBOMB.mkdir(parents=True, exist_ok=True)

# Salvar Partidas
if not df_matches.empty:
    output = BRONZE_STATSBOMB / 'matches.parquet'
    df_matches.to_parquet(output, compression='snappy', index=False)
    
    size_mb = output.stat().st_size / (1024 * 1024)
    
    print(f"✅ Partidas salvas:")
    print(f"   📁 {output}")
    print(f"   📊 {len(df_matches):,} registros")
    print(f"   📦 {size_mb:.2f} MB\n")
    
    logger.info(f"Partidas salvas: {len(df_matches)} registros")

# Salvar Eventos (sample)
if not df_events.empty:
    output = BRONZE_STATSBOMB / 'events_sample.parquet'
    df_events.to_parquet(output, compression='snappy', index=False)
    
    size_mb = output.stat().st_size / (1024 * 1024)
    
    print(f"✅ Eventos (sample) salvos:")
    print(f"   📁 {output}")
    print(f"   📊 {len(df_events):,} registros")
    print(f"   📦 {size_mb:.2f} MB\n")
    
    logger.info(f"Eventos salvos: {len(df_events)} registros")

# Salvar Lineups
if not df_lineups.empty:
    output = BRONZE_STATSBOMB / 'lineups_sample.parquet'
    df_lineups.to_parquet(output, compression='snappy', index=False)
    
    size_mb = output.stat().st_size / (1024 * 1024)
    
    print(f"✅ Lineups (sample) salvos:")
    print(f"   📁 {output}")
    print(f"   📊 {len(df_lineups):,} registros")
    print(f"   📦 {size_mb:.2f} MB\n")
    
    logger.info(f"Lineups salvos: {len(df_lineups)} registros")

# Salvar também CSV (amostra para inspeção)
if not df_matches.empty:
    df_matches.head(100).to_csv(
        BRONZE_STATSBOMB / 'matches_sample.csv',
        index=False,
        encoding='utf-8'
    )
    print(f"✅ Amostra CSV salva: matches_sample.csv (100 primeiras)")

print("\n" + "="*70)
```

```python
print("\n" + "="*70)
print(" "*10 + "🎉 NOTEBOOK 02 - EXTRAÇÃO CONCLUÍDA!")
print("="*70 + "\n")

print("📊 DADOS EXTRAÍDOS:\n")

if not df_matches.empty:
    print(f"✅ PARTIDAS:")
    print(f"   Total: {len(df_matches):,}")
    print(f"   Competições: {df_matches['competition_name'].nunique()}")
    print(f"   Temporadas: {df_matches['season_name'].nunique()}")
    print(f"   Arquivo: data/bronze/statsbomb/matches.parquet\n")

if not df_events.empty:
    print(f"✅ EVENTOS (SAMPLE):")
    print(f"   Total: {len(df_events):,}")
    print(f"   Partidas: {df_events['match_id'].nunique()}")
    print(f"   Tipos de evento: {df_events['type'].nunique()}")
    print(f"   Arquivo: data/bronze/statsbomb/events_sample.parquet\n")

if not df_lineups.empty:
    print(f"✅ LINEUPS (SAMPLE):")
    print(f"   Total: {len(df_lineups):,}")
    print(f"   Jogadores únicos: {df_lineups['player_name'].nunique()}")
    print(f"   Arquivo: data/bronze/statsbomb/lineups_sample.parquet\n")

print("📁 Todos os arquivos salvos em:")
print(f"   📂 {BRONZE_STATSBOMB}/\n")

print("="*70)
print("📌 PRÓXIMO PASSO: Notebook 03 - Data Quality & Cleaning")
print("="*70)

logger.info("Notebook 02 - Extração StatsBomb concluído")
```
